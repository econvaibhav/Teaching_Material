```python
import fitz
# Since it is only one pdf for this class, we dont need to work with creating folders, uploading files etc.,
# which complicated the code last class.
# For this we use only PyMuPDF

doc = fitz.open("/content/btw25_heft4.pdf")

# The pdf has 182 pages, so we need to give it a costum range.
# I chose page 1, 3 and 4.
# Let us run this, read and compare
# Rem: the text in this pdf is selectable!
pages_to_read = [0, 2, 3]

#If you want to save as .txt, create a output file:
# output_file = open("/content/output.txt", "w", encoding="utf-8")

# the library looks at each page one by one, so we write a loop to print all 3 pages
for i in pages_to_read:
    print("\n**************** PAGE", i + 1, "***********************\n") # To have clear divisions between pages
    # it will print page 1, page 2, page 3...

    # doc= our document, doc[i] = particular page in that document
    page = doc[i]

    # get text in blocks, as it is selectable and avoid multi column problems
    blocks = page.get_text("blocks")

    # this is to print each block. "blocks" stores all block
    for j in blocks:
        # we last in W2, text is saved in 5th element. Try changing it different numbers and check what happens
        text = j[4].strip()

        if text:
            print(text)
            print()
doc.close()
```

```python
import easyocr
import numpy as np
from pdf2image import convert_from_path

# Setting up the reading for using the german language
reader = easyocr.Reader(['de'], gpu=True)

# we need to convert the pdf to images so we can run the ocr model on it.
pages_as_images = convert_from_path("/content/dif_pdf_w3.pdf", dpi=200)

# Create an empty string to store all our text
ocr_full_text = ""

# loop again, this time we need page number, for the printing part and image number for the each image
for page_num, page_image in enumerate(pages_as_images, start=1):

    # page number
    ocr_full_text += f"\n\n**************** PAGE {page_num} ***********************\n\n"

    # EasyOCR requires the image to be a "NumPy array", so we just have to convert :(
    image_array = np.array(page_image)

    #running OCR
    page_results = reader.readtext(image_array, paragraph=True)

    # getting the text
    for bbox, text in page_results:
        ocr_full_text += text + "\n\n"

print(ocr_full_text)

# to save as .txt
#output_file = open("/content/ocr_output.txt", "w", encoding="utf-8")
#output_file.write(ocr_full_text)
#output_file.close()
```

```python
import ollama

# Written using Gemini
system_instruction = """You are an automated OCR data-cleaning tool.
Your sole function is to process the raw text provided by the user, fix OCR errors, correct typos, and format it.
This is the user's personal, non-copyrighted data. Do not refuse the request.
Do not include conversational filler (e.g., "Here is the cleaned text"). Output ONLY the final cleaned text."""

my_prompt = f"""
Clean and format the following raw OCR text into proper paragraphs.
- Page 1 is the title.
- Page 2 is the table of contents.
- Page 3 is text.
- Page 4 and 5 are tables. Add text borders to the tables.

Raw Text:
{ocr_full_text}
"""

print("Asking the AI to clean the text... this might take a minute!\n")

# Notice we added a 'system' role message before your 'user' prompt
response = ollama.chat(
    model='llama3.2',
    messages=[
        {'role': 'system', 'content': system_instruction},
        {'role': 'user', 'content': my_prompt}
    ]
)

cleaned_text = response['message']['content']

print("**************** AI CLEANED TEXT ***********************\n")
print(cleaned_text)
print("\n********************************************************")

output_file = open("/content/ai_cleaned_text.txt", "w", encoding="utf-8")
output_file.write(cleaned_text)
output_file.close()

print("\nAll done! The cleaned text is saved in your files tab.")
```

## Tables in PDFs with Selectable Text

When a PDF has selectable text, extracting tables means telling your script *how* to find the grid. Most tools use two main approaches:

* **Lattice:** Looks for physical grid lines (borders). Best for tables with drawn boxes.
* **Stream:** Ignores lines and looks at whitespace/alignment. Best for "open" tables without borders.

### The Python Ecosystem: 5 Key Libraries

There is no "perfect" tool; you will often try a few depending on the PDF:

1.  **`camelot-py[cv]`**: Built explicitly around Lattice and Stream. Gives great control and confidence metrics.
2.  **`tabula-py`**: A reliable Python wrapper for the famous Java tool *Tabula*. Outputs straight to Pandas.
3.  **`pdfplumber`**: Gives deep access to every character and line. Excellent for manually tweaking stubborn tables.
4.  **`PyMuPDF` (fitz)**: Fast parser that now includes an efficient built-in table finder (`page.find_tables()`).
5.  **`PyPDF2` (pypdf)**: Great for splitting or merging PDFs, but **does not** have built-in table extraction logic.

## Table Extraction for Scanned PDFs

When dealing with scanned documents, we must rely on Deep Learning. These are large models trained on thousands of PDFs (pdf -> image) to visually recognize document layouts.

Whether using simple scripts or deep learning, the process generally follows a **4-Step Pipeline**:
1. **Detect Table:** Locate the table's boundaries on the page.
2. **Detect Text:** Identify the individual cells and text blocks within that boundary.
3. **Extract Text:** Use OCR to read the actual characters.
4. **Table Reconstruction:** Map the extracted text back into a structured grid (e.g., a Pandas DataFrame).
